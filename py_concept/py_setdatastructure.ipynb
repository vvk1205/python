```python
s={1,}
s.add(2)
s.add("sam")
s.add("abhi")
print(s)

s2={2,3,4,"abhi", 2.22, 2.414}
print(s.difference(s2)) #difference returns non common value to another set
```

```python
print(s.intersection(s2)) #interection retruns the common value in from set one(left entry)
```

```python
print(s.union(s2))
```
