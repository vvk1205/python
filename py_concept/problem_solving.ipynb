#### reverse string function

```python
flat_string = "this string is for reverse"
rev = ""
for char in flat_string:
    rev = char + rev

print(rev)
```

```python
def reverse_string(enter:str):
    rev=""
    for char in enter:
        rev = char + rev
    return rev
```

```python
reverse_string("this string is for reverse")
```

```python
stirngs= "ethanol"
word=[]
word=list(stirngs.split())
print(word)
vowel=['a','e','i','o','u','A', 'E', 'I', 'O', 'U']
for vow in stirngs:
    if vow in vowel:
        print(True)
    else:
        print(False)
```

```python
def vowel_word(putt:str):
    word=[]
    word= list(putt)
    vowel=['a','e','i','o','u','A', 'E', 'I', 'O', 'U']
    for vow in word:
        if vow in vowel:
            return True
        else:
            return False
```

```python
vowel_word("U")
```

```python
def vowel_word1(putt:str):

    if type(putt) == str and len(putt.strip()) >= 1:  
        firstChar = putt.strip().lower()[0] # converting into lower case for minimize the lookup and taking first index of word
        vowel=('a','e','i','o','u') # changed to tuple for saving some memory space 
        if firstChar in vowel:
            return True
        else:
            return False
    else:
        # return False
        raise Exception("Invalid Input")
       
vowel_word1("1a")
```

```python
vowels= ('a','e','i','o','u')
def vowel_word2(putt:str): 
    if type(putt) == str and len(putt.strip()) >= 1:    
        if putt.strip().lower()[0] in vowels:
            return True
        else:
            return False
    else:
        return False
       
```

```python
vowel_word2("1a")
vowel_word2("a")
```

```python
### identify the vowel also
vowels= ('a','e','i','o','u')
def vowel_word3(putt:str): 
    if type(putt) == str and len(putt.strip()) >= 1:   
        firstChar =  putt.strip().lower()[0]
        if firstChar in vowels:
            print(f"Word is starting with vowel - {firstChar}")
            return True
        else:
            return False
    else:
        return False
       
```

```python
vowel_word3("Unicorn")
```
