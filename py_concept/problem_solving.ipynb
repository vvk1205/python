#### reverse string function

```python
flat_string = "this string is for reverse"
rev = ""
for char in flat_string:
    rev = char + rev

print(rev)
```

```python
def reverse_string(enter:str):
    rev=""
    for char in enter:
        rev = char + rev
    return rev
```

```python
reverse_string("this string is for reverse")
```

#### starting with vowel or not

```python
stirngs= "ethanol"
word=[]
word=list(stirngs.split())
print(word)
vowel=['a','e','i','o','u','A', 'E', 'I', 'O', 'U']
for vow in stirngs:
    if vow in vowel:
        print(True)
    else:
        print(False)
```

```python
def vowel_word(putt:str):
    word=[]
    word=list(putt.split()) 
    vowel=['a','e','i','o','u','A', 'E', 'I', 'O', 'U']
    for vow in word:
        if vow[0] in vowel:
            return True
        else:
            return False
```

```python
vowel_word("hey")
```
