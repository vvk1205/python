### Working with json file

```python
#json.dump(): Serialize obj as a JSON formatted stream to fp(file object)
 
#json.load(): Deserialize fp (a .read()-supporting file-like object containing
            #a JSON document) to a Python object.
```

```python
import json
```

```python
# json.dumps: Serialize obj to a JSON formatted str.
data = ['foo', {'bar': ('baz', None, 1.0, 2)}]
print(type(data))
print(type(json.dumps(data)))
json.dumps(data)
```

```python
# json.dumps: Serialize obj to a JSON formatted str.
data_1 = ['foo', {'bar': ('baz', None, 1.0, 2)}]
print(type(data_1))

create_json_file_name_path=os.path.join(movie_path, "file1.json") 

with open(create_json_file_name_path, mode="w") as jfw: 
    #write method takes only str
    json.dump(data_1, jfw)
 
```

```python
# json.loads: Deserialize s (a str, bytes or bytearray instance
# containing a JSON document) to a Python object.
data1 = '["foo", {"bar": ["baz", null, 1.0, 2]}]'
print(type(data1))
print(type(json.loads(data1)))
json.loads(data1)
```

```python
# json.dumps: Serialize obj to a JSON formatted str.
 
read_json_file_name_path=os.path.join(movie_path, "file1.json") 

with open(read_json_file_name_path, mode="r") as jfw: 
    # load takes file stream and returns the json object reading form the file
    jsonFileContent = json.load(jfw)

print(jsonFileContent)
print(type(jsonFileContent))
```

```python
read_json_movie_data_path=os.path.join(movie_path, "moviedata.json") 

with open(read_json_movie_data_path, mode="r") as jfw: 
    # load takes file stream and returns the json object reading form the file
    movieData = json.load(jfw)
```
