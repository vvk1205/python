#### dictionary

```python
d= {
    "veggies": "potato",
    "fruits": "apple",
    "name": "sky",
    "marks": "100",
    "list": [6,7,9,2,8]
    }
print(d["veggies"])
print(d["fruits"])
d.keys()# it returns only keys(left hand side things)
d.update({"language":"python"}) #it updates the dictionary
d.items()#returns the whole dictionary with keys and values
d.get("veggies") #it returns the value(right hand side things)

```
