#### process management

```python
comp_list=[['c1',0],['c2',0]['c3',0],['c4',0]]
taskDone = []
def busy_comp():
    for singleComp in comp_list:
        for singleTask in task_list :
            if singleComp[1]==0:
                # means free
                singleComp[1] = 1
                # assigning the computer to a task, and prepring the data 
                currentTaskExecution = []
                currentTaskExecution = [singleTask[0], singleTask[1], singleComp[0], str(datetime.now())]
                # processing the task
                time.sleep(singleTask[1])
                # updating the end time
                currentTaskExecution.append( str(datetime.now()))
                # adding the processed task to task done list
                taskDone.append(currentTaskExecution)
                # releasing the computer
                singleComp[1] = 0
                # updating task flag to 1
                singleTask[2] = 1
```

```python
prGraph = 'this is the leasson for python list and it is awesome'
cipherTextList = []
for word in prGraph.split(): # looping through each word
    tempText = ''
    for alpha in word: # looping through each alphabet/char
        currentAlpha = alpha
        if currentAlpha in ['s', 'a', 'w', 'p', 'h']:
            if currentAlpha == 's':
                currentAlpha = '789'
            
            elif currentAlpha == 'a':
                currentAlpha = '158'
            
            elif currentAlpha == 'w':
                currentAlpha = '638'

            elif currentAlpha == 'p':
                currentAlpha = '103'

            elif currentAlpha == 'h':
                currentAlpha = '552'

        tempText = tempText + currentAlpha

    cipherTextList.append(tempText)

' '.join(cipherTextList)
        

            
```

```python
cipherText = 't552i789 i789 t552e le158789789on for 103yt552on li789t 158nd it i789 158638e789ome'
plainTextList = []
for word in cipherText.split(): # looping through each word
    tempText = ''
    tempText = word
    if '789' in tempText:
        tempText = tempText.replace('789', 's')  
    
    if '158' in tempText:
        tempText = tempText.replace('158', 'a') 
    
    if '638' in tempText:
        tempText = tempText.replace('638', 'w') 

    if '103' in tempText:
        tempText = tempText.replace('103', 'p') 

    if '552' in tempText:
        tempText = tempText.replace('552', 'h') 

    
    plainTextList.append(tempText)

' '.join(plainTextList)
        

            
```
