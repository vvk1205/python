#### checking duplicate entries in list

```python
n=0
dup=[]
a=len(lis)

for i in range(0, a):
    count=0
    for j in range(0, a):
    
        if lis[i]== lis[j]:
            count=count+1
    if count>1:
        dup.append(lis[i])
print("duplicate items:", dup)        

        
```

#### appendind list into list

```python
details=[]
for i in range(0, len(dup)):
    matched=[]

    for j in range(0, a):
        
        if dup[i]==lis[j]:
            
            matched.append(j)
    v = [dup[i]]
    v.extend(matched)
    details.append(v)
print("the detail is: ", details )
    

```
