```python
try:
    print(empDict["nam"])
except Exception as e: # Exception is a base exception
    print(e) 
    print("key not found")
```

```python
try:
    print(empDict["nam"])
except KeyError as ke: # KeyError is sub exception
    print(ke) 
    print("key not found")
```

```python
employee_dict= {
                    "name": "john C",
                    "age": 39,
                    "salary": "$210"
    
                    }
emp_name=""
for k,v in employee_dict.items():
    if k== "name":
        emp_name=v
        print(f"employee name is:",v)
    if k== "age":
        print(f"{emp_name} is {v} years old")
    if k=="salary":        
        print(f"{emp_name} earns {v} in a month")
```

#### program to print max salary oldest wrestler min salary and youngest with min salary

```python
emp_datalist=[{
            "name": "john",
            "age": 39,
            "salary": 6000
           
            },
            {
            "name": "undertaker",
            "age": 51,
            "salary": 8000
            },
            {
            "name": "triple H",
            "age": 49,
            "salary": 3000
        
            },
            {
            "name": "bigshow",
            "age": 52,
            "salary": 5000
           
            },
            {
            "name": "rock",
            "age": 45,
            "salary": 11000
           
            },
            {
            "name": "batista",
            "age": 42,
            "salary": 9000
            }]
sal=[]
for emp_data in emp_datalist:
    for k,v in emp_data.items():
        if k== "salary":
            sal.append(v)
ageList=[]
for emp_data in emp_datalist:
    for k,v in emp_data.items():
        if k== "age":
            ageList.append(v)
print(ageList)
print(min(ageList))
max_age= ageList[0]
for highest_age in ageList: #oldest
    if highest_age > max_age:    
        max_age = highest_age
print(f"the oldest wrestler of the employee dictionary is:",max_age)

young= ageList[0]
for youngest_age in ageList: #youngest
    if youngest_age < young:    
        young = youngest_age
print(f"the youngest wrestler of the employee dictionary is:",young)

max = sal[0]
for max_sal in sal: #maximum salary
    if max_sal > max:    
        max = max_sal
print(f"the maximum salary of the employee dictionary is:",max) 

mini= sal[0]
for mini_sal in sal: #minimum salary
    if mini_sal < mini:    
        mini = mini_sal
print(f"the minimum salary of the employee dictionary is:",mini)
print(f"youngest age of dictionary is: {young} and maximum salary is: {max}")
```
