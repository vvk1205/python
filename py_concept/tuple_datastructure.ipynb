#### check vowel in sentence

```python
a = ['hey','ae', 'i', 'am', 'vivek', 'singh', 'and', 'i', 'am', 'a', 'learner', 'of', 'python', 'and', 'here', 'i', 'am', 'creating', 'a', 'list']

print(a) 
vowelss = []
voweltup = ('a','e','i','o','u')
count=0
for i in a:
    if i[0] in voweltup:
        count=count+1
        vowelss.append(i)
print(count)        
# Printing result
print("list of words starting with vowels : ",(vowelss))
print(sys.getsizeof(vowelss))
```
